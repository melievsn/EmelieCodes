```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df = pd.read_csv("all_recs.csv", encoding= "unicode_escape")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
year_order = df['Year'].value_counts().index
location_order = df['Location'].value_counts().index
motive_order = df['Motive'].value_counts().index
industry_order = df['Industry'].value_counts().index
```

```python
plt.figure(figsize=(20,8))
plt.title("Attack Frequency by Year")
sb.countplot(data = df, y = 'Year', color = sb.color_palette()[2], order = year_order)
plt.xticks(rotation = 45);
```

```python
plt.figure(figsize=(20,8))
plt.title("Attack Frequency by Motive")
sb.countplot(data = df, y = 'Motive', color = sb.color_palette()[5], order = motive_order)
plt.xticks(rotation = 45);
```

```python
plt.figure(figsize=(20,48))
plt.title("Attack Frequency by Location")
sb.countplot(data = df, y = 'Location', color = sb.color_palette()[7], order = location_order)
plt.xticks(rotation = 45);
```

```python
plt.figure(figsize=(20,8))
plt.title("Attack Frequency by Industry")
sb.countplot(data = df, y = 'Industry', color = sb.color_palette()[3], order = industry_order)
plt.xticks(rotation = 45);
```
